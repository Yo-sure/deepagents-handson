이 Agent는 자료 검색 → 본문 읽기 → 근거를 연결한 설명을 수행합니다. DeepAgents의 계획·하위 작업 기능도 사용할 수 있지만 모든 질문에 쓰도록 강제하지 않습니다. 실행 결과를 대신 검사하는 채점기는 아닙니다. 답변의 근거를 직접 열고 자신의 출력과 비교하십시오.


```python
from pathlib import Path
import os
import sys

root = Path.cwd()
if not (root / "course").is_dir():
    root = root.parent
if not (root / "course" / "common.py").is_file():
    raise RuntimeError("workshop/notebooks에서 노트북을 여십시오.")
os.chdir(root)
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
from course.qna import build_qna_agent
qna_agent = build_qna_agent()
print("준비 완료. 아래 질문만 바꾸어 실행하십시오.")
```

```python
question = "2B에서 visited를 ['ask']로 바꾸면 무엇이 달라지나요? 풀이 코드와 출력 해석을 근거로 설명해주세요."

result = qna_agent.invoke(
    {"messages": [{"role": "user", "content": question}]},
    config={"recursion_limit": 35},
)
from IPython.display import Markdown, display
display(Markdown(str(result["messages"][-1].content)))
```
